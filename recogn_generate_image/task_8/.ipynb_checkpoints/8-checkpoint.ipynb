```python
# ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿
'''
CIFAR-10 ﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿ 60 000 ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ 32﷿﷿32 ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿, 10 ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿
(﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿, ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿, ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿, ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿, ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿, ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿, ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿, ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿, ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿, ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿).
'''
# ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿: ﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ~170 ﷿﷿﷿﷿ ﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿!
(train_images, train_labels), (test_images, test_labels) = datasets.cifar10.load_data()
train_images, test_images = train_images / 255.0, test_images / 255.0

# ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿: validation_split ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ 10% ﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿
# ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿, ﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿
idx = np.random.permutation(len(train_images))
train_images, train_labels = train_images[idx], train_labels[idx]

class_names = ['airplane', 'automobile', 'bird', 'cat', 'deer',
               'dog', 'frog', 'horse', 'ship', 'truck']

plt.figure(figsize=(8, 8))
for i in range(9):
    plt.subplot(3, 3, i + 1)
    plt.imshow(train_images[i])
    plt.title(class_names[train_labels[i][0]])
    plt.axis('off')
plt.show()
```

```python
# ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ CNN:
# Conv -> Pool -> Conv -> Pool -> Conv -> Flatten -> Dense -> Dense
# ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿: ﷿﷿﷿﷿﷿﷿﷿﷿ Input ﷿﷿ Conv2D ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ (input_shape ﷿﷿ Conv2D ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿)

model = models.Sequential([
    layers.Input(shape=(32, 32, 3)),              # ﷿﷿﷿﷿﷿﷿﷿﷿: ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ 32﷿﷿32
    layers.Conv2D(32, (3, 3), activation='relu'), # 32 ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ 30﷿﷿30
    layers.MaxPooling2D((2, 2)),                  # ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿ 15﷿﷿15
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.Flatten(),                             # 4﷿﷿4﷿﷿64 = 1024 ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿
    layers.Dense(64, activation='relu'),
    layers.Dense(10)                              # ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ (softmax ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿)
])
model.summary()
```

### ﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿: train / validation / test

﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿:

* **train** ﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿;
* **validation** ﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ (﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿, early stopping);
* **test** ﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿, ﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿.

﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿: ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿
﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿, ﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿.
﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿ 10% ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ (`validation_split=0.1`),
﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿.

```python
model.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
              metrics=['accuracy'])

# ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿ ﷿﷿﷿﷿ 10% ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿; ﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿
history = model.fit(train_images, train_labels,
                    epochs=10,
                    validation_split=0.1,
                    verbose=2)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4))

ax1.plot(history.history['loss'], label='Train loss')
ax1.plot(history.history['val_loss'], label='Val loss')
ax1.set_title('﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿')
ax1.set_xlabel('﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿')
ax1.legend()

ax2.plot(history.history['accuracy'], label='Train accuracy')
ax2.plot(history.history['val_accuracy'], label='Val accuracy')
ax2.set_title('﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿')
ax2.set_xlabel('﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿')
ax2.legend()

plt.show()

test_loss, test_acc = model.evaluate(test_images, test_labels, verbose=0)
print(f"\n﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ (﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿): {test_acc:.4f}")
```

### ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿

*﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿: `tf.keras.preprocessing.image.*` ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿ ﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ Keras ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ `tf.keras.utils.load_img` / `img_to_array`.*

```python
img = tf.keras.utils.load_img('horse.jpg', target_size=(32, 32))

img_array = tf.keras.utils.img_to_array(img)
img_array = img_array / 255.0                # ﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿
img_array = np.expand_dims(img_array, axis=0)  # ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿: (1, 32, 32, 3)

plt.imshow(img)
plt.title('﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ (﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿ 32﷿﷿32)')
plt.axis('off')
plt.show()
```

```python
predictions = model.predict(img_array, verbose=0)

# ﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿ Softmax (﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿ from_logits=True) ﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿,
# ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿
probabilities = tf.nn.softmax(predictions[0]).numpy()

# ﷿﷿﷿﷿﷿﷿﷿﷿ argmax ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿-3
top3_idx = np.argsort(probabilities)[::-1][:3]
print("﷿﷿﷿﷿﷿﷿-3 ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿:")
for i in top3_idx:
    print(f"  {class_names[i]:12s} p = {probabilities[i]:.2f}")

# ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ (﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿ ﷿﷿﷿﷿﷿﷿-3)
plt.figure(figsize=(9, 4))
bars = plt.bar(range(10), probabilities, color='steelblue')
for i in top3_idx:
    bars[i].set_color('orange')
plt.xticks(range(10), class_names, rotation=45)
plt.title('﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ Softmax: ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿ ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿')
plt.tight_layout()
plt.show()
```

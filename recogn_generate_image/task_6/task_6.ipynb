**Тема:** Классификация изображений с использованием классических методов машинного обучения

---

## 1. Постановка задачи классификации изображений

**Классификация изображения** – это процесс отнесения входного изображения к одному из заранее определённых классов.

Примеры задач:

* Определение рукописной цифры (0–9) на картинке.
* Распознавание букв или символов.
* Классификация объектов: кошка / собака / автомобиль.
* Определение типа ткани на медицинском изображении.

**Схема решения задачи классификации:**

1. Сбор и разметка набора данных.
2. Предобработка изображений (масштабирование, нормализация, перевод в градации серого).
3. Выделение признаков (например, HOG, ключевые точки, или просто «сырые пиксели»).
4. Обучение классификатора.
5. Оценка качества (accuracy, confusion matrix).

---

## 2. Подходы к классификации изображений

Существует два основных подхода:

1. **Классический (машинное обучение)**

   * Используются заранее определённые признаки (ручная инженерия признаков).
   * Алгоритмы: kNN, SVM, Random Forest и др.
   * Плюсы: простота реализации, работает на небольших данных.
   * Минусы: зависимость от выбора признаков, ограниченная гибкость.

2. **Современный (нейросети, CNN)**

   * Признаки извлекаются автоматически обучаемой моделью.
   * Более высокая точность, но требует много данных и вычислительных ресурсов.

На данном занятии мы фокусируемся на **классических методах**.

---

## 3. Методы классификации изображений (классический ML)

### 🔹 3.1. k-Nearest Neighbors (kNN). Метод ближайших соседей

* Один из самых простых алгоритмов.
* Для нового изображения алгоритм ищет **k ближайших объектов** в обучающей выборке (по метрике расстояния, например, Евклидовой).
* Класс присваивается по большинству голосов соседей.



**Особенности:**

* Не требует обучения (данные просто запоминаются).
* Чувствителен к выбору метрики расстояния и значению k.
* Медленный при классификации больших наборов данных.

Применение: распознавание рукописных цифр (MNIST), простые задачи.

статья https://habr.com/ru/articles/680004/

---

### 🔹 3.2. Support Vector Machine (SVM). Метод опорных векторов

* Алгоритм строит **разделяющую гиперплоскость** между классами.
* Идея: найти такую границу, которая максимально отделяет один класс от другого (максимизирует отступ).
* Для сложных случаев применяются **ядровые функции** (например, RBF-ядро), позволяющие разделять данные в более высоких измерениях.



**Особенности:**

* Хорошо работает при малом числе признаков.
* Часто используется для распознавания лиц, текста, биометрии.
* Минусы: медленный на больших выборках, требует настройки параметров.

статья https://habr.com/ru/companies/skillfactory/articles/857162/

---

### 🔹 3.3. Random Forest. Случайный лес

* Ансамблевый метод, основанный на деревьях решений.
* Каждое дерево строится на случайном подмножестве данных и признаков.
* Итоговый класс определяется голосованием всех деревьев.



**Особенности:**

* Высокая точность «из коробки», редко переобучается.
* Устойчив к шумам.
* Работает быстро даже на больших выборках.
* Минус: менее интерпретируемый по сравнению с одним деревом решений.

статья https://sysblok.ru/glossary/chto-takoe-random-forest/

---

## 4. Подготовка данных для классификации

1. **Нормализация изображений:**

   * приведение к единому размеру (например, 28×28 для MNIST);
   * преобразование в оттенки серого;
   * нормализация значений пикселей (0–1 или -1…1).

2. **Извлечение признаков:**

   * Сырые пиксели (flatten → вектор).
   * HOG (гистограммы градиентов).
   * PCA (снижение размерности).

3. **Формирование обучающей и тестовой выборок** (train/test split).

---

## 5. Сравнение методов

| Метод             | Принцип работы                  | Преимущества                          | Недостатки                         | Применение                       |
| ----------------- | ------------------------------- | ------------------------------------- | ---------------------------------- | -------------------------------- |
| **kNN**           | Сравнение с ближайшими соседями | Простота, нет обучения                | Медленный, чувствителен к шуму     | Малые наборы данных              |
| **SVM**           | Разделяющая гиперплоскость      | Высокая точность, устойчивость        | Долго обучается, требует настройки | Распознавание лиц, текста        |
| **Random Forest** | Ансамбль деревьев               | Быстрота, устойчивость, мало настроек | Менее интерпретируем               | Классификация объектов, медицина |

---

## 6. Применение в реальных задачах

* **kNN**: быстрое прототипирование, биометрические системы.
* **SVM**: распознавание лиц, текста, контроль качества на производстве.
* **Random Forest**: классификация медицинских изображений, спутниковые снимки, анализ дефектов.

---

## 7. Основные выводы

1. Классификация изображений — базовая задача компьютерного зрения.

4. В реальных системах классические методы часто служат базовой моделью или частью более сложного пайплайна.

```python
import os
import urllib.request
import numpy as np

MNIST_PATH = "mnist.npz"
if not os.path.exists(MNIST_PATH):
    urllib.request.urlretrieve(
        "https://storage.googleapis.com/tensorflow/tf-keras-datasets/mnist.npz", MNIST_PATH
    )


class MNIST:
    @staticmethod
    def load_data():
        with np.load(MNIST_PATH) as d:
            return (d["x_train"], d["y_train"]), (d["x_test"], d["y_test"])


mnist = MNIST()
```

```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt

digits = load_digits()
print("Размерность данных:", digits.data.shape)
print("Классы:", digits.target_names)

plt.gray()
plt.matshow(digits.images[0])
plt.title(f"Метка: {digits.target[0]}")
plt.show()
```

```python
n_rows = 6
_, axes = plt.subplots(nrows=n_rows, ncols=10, figsize=(12, n_rows * 1.5))

for ax, image, label in zip(axes.flatten(), digits.images, digits.target):
    ax.set_axis_off()
    ax.imshow(image, cmap=plt.cm.gray_r, interpolation="nearest")
    ax.set_title("Training: %i" % label)

plt.tight_layout()
```

```python
# классификация KNN

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

X_train, X_test, y_train, y_test = train_test_split(
    digits.data, digits.target, test_size=0.3, random_state=42
)

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)

y_pred = knn.predict(X_test)
print("Точность kNN:", accuracy_score(y_test, y_pred))
```

```python
from sklearn.svm import SVC

svm = SVC(kernel='linear')
svm.fit(X_train, y_train)

y_pred = svm.predict(X_test)
print("Точность SVM:", accuracy_score(y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)

y_pred = rf.predict(X_test)
print("Точность Random Forest:", accuracy_score(y_test, y_pred))
```

### Оценка качества: матрица ошибок (confusion matrix)

Одна точность (accuracy) не показывает, *где* модель ошибается.
Матрица ошибок сравнивает истинные классы (строки) с предсказанными (столбцы):
диагональ — правильные ответы, всё вне диагонали — ошибки.

```python
# Строим матрицу ошибок для kNN — лучшей из трёх моделей.
# Для остальных классификаторов — аналогично (SVM: svm.predict, RF: rf.predict).

from sklearn.metrics import ConfusionMatrixDisplay

y_pred_knn = knn.predict(X_test)  # knn уже обучен выше

disp = ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred_knn, cmap="Blues", colorbar=False
)
disp.ax_.set_title("Матрица ошибок kNN на тестовой выборке digits")
plt.show()

# Вопросы для обсуждения:
# 1) Какие цифры чаще всего путаются? (ищите яркие ячейки ВНЕ диагонали)
# 2) Каких ошибок больше: у «похожих» цифр (3↔8, 9↔5, 4↔9) или случайных?
```

```python
import time
import cv2
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
def show(images, titles, figsize=(16, 4)):
    fig, ax = plt.subplots(1, len(images), figsize=figsize)
    for a, im, t in zip(np.atleast_1d(ax), images, titles):
        a.imshow(im, cmap="gray")
        a.set_title(t)
        a.axis("off")
    plt.tight_layout()
    plt.show()


def predict_digit(path, model):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
    img28 = cv2.resize(img, (28, 28))
    img28 = 255 - img28
    x = (img28 / 255.0).reshape(1, -1)
    scores = model.decision_function(x)[0]
    return img, img28, int(np.argmax(scores)), scores


def top_confusions(cm, n=5):
    pairs = [(cm[i, j], i, j) for i in range(10) for j in range(10) if i != j and cm[i, j] > 0]
    return sorted(pairs, reverse=True)[:n]
```

### Задание 1. Перебор `n_neighbors` для kNN и график accuracy(k)

```python
X_tr, X_te, y_tr, y_te = train_test_split(
    digits.data, digits.target, test_size=0.3, random_state=42
)

ks = list(range(1, 16))
acc_train, acc_test = [], []
for k in ks:
    model = KNeighborsClassifier(n_neighbors=k).fit(X_tr, y_tr)
    acc_train.append(accuracy_score(y_tr, model.predict(X_tr)))
    acc_test.append(accuracy_score(y_te, model.predict(X_te)))

best_k = ks[int(np.argmax(acc_test))]
print(" k | обучение |  тест")
for k, a, b in zip(ks, acc_train, acc_test):
    print(f"{k:2d} |   {a:.3f}   | {b:.3f}" + ("   -- лучшее k" if k == best_k else ""))
print(f"Оптимальное k: {best_k}")

plt.figure(figsize=(9, 4))
plt.plot(ks, acc_train, "o-", label="обучение")
plt.plot(ks, acc_test, "s-", label="тест")
plt.axvline(best_k, color="red", ls="--", label=f"лучшее k = {best_k}")
plt.xticks(ks)
plt.xlabel("k")
plt.ylabel("Accuracy")
plt.title("Обучение kNN от 1 до 15")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

### Задание 2. SVM с ядрами `linear` и `rbf` на digits

```python
results = {}
for kernel in ("linear", "rbf"):
    t0 = time.time()
    svm = SVC(kernel=kernel).fit(X_tr, y_tr)
    results[kernel] = accuracy_score(y_te, svm.predict(X_te))
    print(f"SVM ({kernel}): accuracy = {results[kernel]:.4f}, обучение {time.time() - t0:.1f} с")

print(f"Разница в точности: {(results['rbf'] - results['linear']) * 100:+.2f} п.п.")

plt.bar(results.keys(), results.values(), color=["tab:blue", "tab:orange"])
plt.ylim(0.7, 1.0)
plt.ylabel("Accuracy")
plt.title("SVM с разными ядрами на digits")
plt.show()
```

## Тема: Введение в дисциплину. Основы работы с изображениями в Python

---

### 1. Что такое компьютерное зрение и распознавание изображений?

**Компьютерное зрение (Computer Vision, CV)** – это область искусственного интеллекта, которая занимается анализом и пониманием изображений и видео.
Основная задача: научить компьютер «видеть» и интерпретировать визуальную информацию так, как это делает человек.

**Распознавание изображений** – одна из ключевых задач CV, связанная с классификацией, выделением объектов, анализом сцен.

#### Примеры применения:

* **Медицина** – автоматический анализ рентгеновских снимков и МРТ.
* **Автопром** – системы компьютерного зрения в беспилотных автомобилях.
* **Безопасность** – системы распознавания лиц и номеров автомобилей.
* **Индустрия развлечений** – дополненная и виртуальная реальность, фильтры в приложениях.

---

### 2. Основные задачи распознавания изображений

1. **Классификация изображений**

   * Определение, к какому классу относится всё изображение (например: «кот» или «собака»).

2. **Детекция объектов**

   * Нахождение объектов на изображении и определение их координат (рамки вокруг объектов).

3. **Сегментация**

   * Разделение изображения на области (например, выделение дороги, машин и пешеходов на фото).

4. **Отслеживание объектов (Tracking)**

   * Слежение за движением объекта в видеопотоке.

---

### 3. Этапы обработки изображений

Для того чтобы распознавание было успешным, обычно выделяют несколько шагов:

1. **Предобработка**

   * коррекция яркости и контраста;
   * удаление шумов;
   * преобразование к нужному размеру или цветовой модели.

2. **Выделение признаков**

   * преобразование изображения в набор числовых характеристик (например: углы, границы, текстуры).

3. **Классификация / принятие решения**

   * применение алгоритмов машинного обучения или нейронных сетей для распознавания.

---

### 4. Представление изображения в компьютере

#### 4.1. Цифровое изображение

* Изображение в компьютере — это **матрица пикселей**.
* Каждый пиксель хранит числовые значения (интенсивность света или цвет).

#### 4.2. Градации серого (Grayscale)

* Изображение описывается одним числом на пиксель:

  * `0` = чёрный,
  * `255` = белый.

#### 4.3. Цветные изображения (RGB)

* Каждый пиксель описывается тремя числами: `[R, G, B]`.
* Например: `[255, 0, 0]` = красный цвет, `[0, 255, 0]` = зелёный.

#### 4.4. Размер изображения

* Определяется количеством строк (высота), столбцов (ширина) и числом каналов (1 для grayscale, 3 для RGB).
* Пример: изображение `640×480×3` → 640 пикселей по ширине, 480 по высоте, 3 канала цвета.

```python
# Загрузка и отображение изображений

import cv2
import matplotlib.pyplot as plt

# Загрузка изображения
img = cv2.imread('example.jpg')  # по умолчанию BGR
print("Размер:", img.shape)  # (высота, ширина, каналы)

# Отображение в OpenCV (BGR)
cv2.imshow("Image (OpenCV)", img)
cv2.waitKey(0)
cv2.destroyAllWindows()

# Отображение в matplotlib (нужно конвертировать в RGB)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
plt.title("Image (Matplotlib)")
plt.axis("off")
plt.show()
```

```python
# Наглядная демонстрация BGR vs RGB
# OpenCV хранит каналы в порядке B-G-R, а matplotlib ожидает R-G-B.
# Если забыть про конвертацию, красный и синий цвета "поменяются местами".

fig, ax = plt.subplots(1, 2, figsize=(12, 6))

# Слева: без конвертации (ошибка!) — цвета перепутаны
ax[0].imshow(img)
ax[0].set_title("Без конвертации (BGR) — цвета перепутаны")
ax[0].axis("off")

# Справа: после конвертации (правильно)
ax[1].imshow(img_rgb)
ax[1].set_title("После cvtColor (RGB) — правильно")
ax[1].axis("off")

plt.show()
```

```python
# Представление изображения как массива

print("Тип данных:", type(img))
print("Форма массива:", img.shape)
print("Пример значения пикселя [100,100]:", img[100,100])  # [B, G, R]
```

```python
# для grayscale-матрицы обязательно указывайте cmap="gray" в plt.imshow.
# Без него matplotlib применяет cmap по умолчанию  и раскрашивает чёрно-белую картинку в зелёно-фиолетовые цвета

# Конвертация в оттенки серого

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.show()
```

```python
# Сохранение изображения

cv2.imwrite("output_gray.jpg", gray)
```

### Change red and blue channel

```python
def change_r_b(image_change: np.ndarray) -> np.ndarray:
    return image_change[:, :, [2, 1, 0]]

img_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
image_changed = change_r_b(img_rgb.copy())

fig, ax = plt.subplots(1, 2, figsize=(12, 6))

ax[0].imshow(img_rgb)
ax[0].set_title("Orig")
ax[0].axis("off")

ax[1].imshow(image_changed)
ax[1].set_title("Change red and blue channel")
ax[1].axis("off")

plt.show()
```

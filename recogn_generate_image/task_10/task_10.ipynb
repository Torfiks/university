## Конвертация PDF в Markdown при помощи MinerU

Запусти следующую ячейку: все страницы `mil.pdf` сохранятся в `mil.md` рядом с PDF.
Используем локальный режим `flash`: для `standard` в текущей конфигурации отключён движок.
Документ не отправляется в облако. Качество распознавания и оформления может отличаться от `standard`.

Команды: `mineru server start`, затем `mineru parse mil.pdf --pages all -o mil.md --tier flash --wait 1800`.


````python
import subprocess
import sys
from pathlib import Path

input_path = Path("/home/roma/PycharmProjects/university/recogn_generate_image/task_10/mil.pdf")
output_path = input_path.with_suffix(".md")
if not input_path.is_file():
    raise FileNotFoundError(f"Документ не найден: {input_path}")

# Используем MinerU из окружения ядра ноутбука.
mineru = [sys.executable, "-c", "from mineru.cli.main import main; main()"]
subprocess.run([*mineru, "server", "start"], check=True)
subprocess.run(
    [*mineru, "parse", str(input_path), "--pages", "all",
     "--tier", "flash", "-o", str(output_path), "--wait", "1800"],
    check=True,
)
if not output_path.is_file():
    raise RuntimeError(f"MinerU не создал файл: {output_path}")
print(f"Markdown сохранён: {output_path}")
````

Проблемы, которые нужно исправить:
- `Date` — хранится как текст, смесь форматов `%Y-%m-%d` и `%d.%m.%Y`;
- `SalesAmount` — текст, часть значений начинается с `$`;
- `SalesAmount`, `Discount`, `Quantity` — нет пропусков и отрицательных кол-в кроме Quantity;
- `Quantity` — есть отрицательные значения;
- `Discount` — есть пропуски (NaN);
- `ProductCategory` — лишние пробелы и разнобой регистра (`Books`, `ELECTRONICS`, `Home `, ` Clothing `);
- `Region` — разнобой регистра (`East`/`east`);
- есть полностью дублирующиеся строки и дубликаты по `TransactionID`.

### Задание 2. Очистка текстовых столбцов

```python
df['ProductCategory'] = df['ProductCategory'].str.strip()
df['ProductCategory'] # убрать пробелы по краям
```

```python
df['ProductCategory'] = df['ProductCategory'].str.lower()
df['ProductCategory'] # в нижний регистр
```

```python
df['Region'] = df['Region'].str.strip().str.lower().str.title()
df['Region']
```

### Задание 3. Исправление форматов чисел и дат

```python
df["SalesAmount"] = df["SalesAmount"].replace(r"\$", "", regex=True)
df["SalesAmount"] = df["SalesAmount"].astype('float')
df["SalesAmount"]
```

```python
df["Date"] = pd.to_datetime(df["Date"], format="mixed", dayfirst=True)
df["Date"]
```

```python
df["Quantity"] = df["Quantity"].abs()
df["Quantity"]
```

### Задание 4. Обработка пропусков

```python
df['Discount'] = df['Discount'].fillna(df['Discount'].mean())
df['Discount']
```

```python
df['SalesAmount'] = df['SalesAmount'].fillna(df['SalesAmount'].mean())
df['SalesAmount']
```

```python
# Пропусков в Quantity нет (см. Задание 1), чистить не нужно.
# Удалим строки с пропусками в критичных столбцах (если появятся):
df = df.dropna(subset=['SalesAmount', 'Quantity', 'Discount'])
```

### Задание 5. Удаление дубликатов

```python
df = df.drop_duplicates()
df
```

```python
df = df.drop_duplicates(subset=['TransactionID'])
df
```

```python
df.to_csv('clean_sales.csv', index=False)
print(f"Сохранено строк: {df.shape[0]}, пропусков: {df.isnull().sum().sum()}")
```
